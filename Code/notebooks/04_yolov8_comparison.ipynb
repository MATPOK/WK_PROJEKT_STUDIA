```python
#Kod do wykrywania obiektów za pomocą modelu YOLOv8-nano
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import os

model = YOLO("yolov8n.pt") 

# Folder bazowy z zdjęciami testowymi
base_path = '../data/test_images/'

# Lista plików do przetworzenia
image_names = ['1.jpg', '2.jpg', '3.jpg', '4.jpg', '5.jpg']

for img_name in image_names:
    image_path = os.path.join(base_path, img_name)
    
    if not os.path.exists(image_path):
        print(f"\nNie znaleziono pliku: {image_path}")
        continue
    
    results = model(image_path)

    # Zdjęcie z gotowymi ramkami
    res_plotted = results[0].plot()

    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(res_plotted, cv2.COLOR_BGR2RGB))
    plt.title(f"YOLOv8-nano - {img_name}")
    plt.axis('off')
    plt.show()
```
