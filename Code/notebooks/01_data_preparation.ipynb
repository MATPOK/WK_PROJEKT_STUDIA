```python
# Przygotowanie danych: wybor 5 klas z bazy GTSRB i skopiowanie ich do osobnego folderu
import os
import shutil
import matplotlib.pyplot as plt
import cv2

source_dir = '../data/Train'
target_dir = '../data/selected_signs'

selected_classes = {
    1:  'Ograniczenie_30',
    12: 'Droga_z_pierwszenstwem',
    13: 'Ustap_pierwszenstwa',
    14: 'Stop',
    17: 'Zakaz_wjazdu',
}

os.makedirs(target_dir, exist_ok=True)

for class_id, class_name in selected_classes.items():
    source_class_dir = os.path.join(source_dir, f"{class_id:05d}")
    if not os.path.exists(source_class_dir):
        source_class_dir = os.path.join(source_dir, str(class_id))
    target_class_dir = os.path.join(target_dir, class_name)
    os.makedirs(target_class_dir, exist_ok=True)

    if os.path.exists(source_class_dir):
        images = [f for f in os.listdir(source_class_dir) if f.endswith(('.ppm', '.png', '.jpg'))]
        for img_name in images:
            shutil.copy2(os.path.join(source_class_dir, img_name),
                         os.path.join(target_class_dir, img_name))
        print(f"Skopiowano {len(images)} zdjec dla klasy: {class_name}")
    else:
        print(f"Nie znaleziono folderu zrodlowego dla klasy {class_name} ({class_id})")

print("Zakonczono kopiowanie zdjec.")
```

```python
# Podglad: po jednym przykladowym zdjeciu z kazdej wybranej klasy
fig, axes = plt.subplots(1, 5, figsize=(20, 5))
for ax, class_name in zip(axes, selected_classes.values()):
    class_path = os.path.join(target_dir, class_name)
    if os.path.exists(class_path):
        img = cv2.imread(os.path.join(class_path, os.listdir(class_path)[0]))
        ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        ax.set_title(class_name)
        ax.axis('off')
plt.show()
```
